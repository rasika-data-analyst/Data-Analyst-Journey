# NumPy Arrays
A NumPy array is a powerful data structure in Python used for numerical computations. It is provided by the NumPy (Numerical Python) library and is more efficient than Python lists for handling large datasets.

* Homogeneous – All elements in a NumPy array must be of the same data type (e.g., integers, floats).
* Fast and Memory Efficient – Uses contiguous memory blocks, making operations faster than lists.
* Supports Vectorized Operations – Element-wise operations can be performed without loops.
* Multi-Dimensional – Can handle 1D (vector), 2D (matrix), and n-D arrays.
* Convenient Functions – Includes built-in functions for mathematical operations, statistics, and linear algebra.

# Difference between Python List and NumPy Array

Lists are built-in Python structures that can store mixed data types, but are slower and don’t support vectorized mathematical operations.

NumPy Arrays store homogeneous data, are faster and memory-efficient, and support vectorized operations and multi-dimensional data structures — ideal for numerical and scientific computing.

# NumPy for Data Analysts – Additional Practice

This section extends the existing NumPy array practice with topics that are especially useful for Data Analyst work: missing values, unique/frequency analysis, percentiles, conditional classification, clipping, and business-style numerical analysis.

## 1. Missing Values with NumPy

NumPy represents missing numerical values using `np.nan`. The `nan*` functions allow calculations while ignoring missing values.

```python
import numpy as np

sales = np.array([1200, 1500, np.nan, 1800, 2100, np.nan, 2500])

print("Missing values:", np.isnan(sales))
print("Number of missing values:", np.isnan(sales).sum())
print("Mean ignoring missing values:", np.nanmean(sales))
print("Median ignoring missing values:", np.nanmedian(sales))
print("Sum ignoring missing values:", np.nansum(sales))
```

## 2. Unique Values and Frequency Analysis

`np.unique()` is useful for identifying distinct values and their frequencies.

```python
categories = np.array(["Laptop", "Mobile", "Laptop", "Tablet", "Mobile", "Laptop", "Tablet"])

print("Unique categories:", np.unique(categories))
values, counts = np.unique(categories, return_counts=True)

for value, count in zip(values, counts):
    print(value, ":", count)
```

## 3. Percentiles and Quantiles

Percentiles help understand the distribution of numerical business metrics such as sales, salary, transaction value, or delivery time.

```python
sales = np.array([1200, 1500, 1800, 2100, 2500, 2800, 3200, 3500, 4200, 5000])

print("25th percentile:", np.percentile(sales, 25))
print("50th percentile / median:", np.percentile(sales, 50))
print("75th percentile:", np.percentile(sales, 75))
print("90th percentile:", np.percentile(sales, 90))
print("0.25 quantile:", np.quantile(sales, 0.25))
```

## 4. Conditional Classification with `np.where()`

Use vectorized conditions to classify numerical data without writing a Python loop.

```python
sales = np.array([800, 1500, 2300, 3100, 4500])

sales_category = np.where(sales >= 2500, "High", "Low")
print(sales_category)
```

## 5. Multiple Conditions with `np.select()`

`np.select()` is useful when a business metric needs more than two categories.

```python
sales = np.array([500, 1200, 2200, 3500, 5000])

conditions = [
    sales < 1000,
    (sales >= 1000) & (sales < 3000),
    sales >= 3000
]

choices = ["Low", "Medium", "High"]
sales_category = np.select(conditions, choices, default="Unknown")

print(sales_category)
```

## 6. Limiting Values with `np.clip()`

`np.clip()` limits values to a specified minimum and maximum. This can be useful when handling extreme values for analytical exercises.

```python
discount = np.array([-5, 0, 10, 25, 110])

clean_discount = np.clip(discount, 0, 100)
print("Original:", discount)
print("Clipped:", clean_discount)
```

## 7. Axis-Based Analysis

For a 2-D dataset, `axis=0` performs calculations down rows for each column, while `axis=1` performs calculations across columns for each row.

```python
monthly_sales = np.array([
    [1200, 1500, 1800],
    [1300, 1600, 1900],
    [1400, 1700, 2000]
])

print("Column totals:", np.sum(monthly_sales, axis=0))
print("Row totals:", np.sum(monthly_sales, axis=1))
print("Column averages:", np.mean(monthly_sales, axis=0))
print("Row averages:", np.mean(monthly_sales, axis=1))
```

## 8. Business-Style NumPy Analysis

The following example combines NumPy concepts to answer common Data Analyst questions.

```python
sales = np.array([1250, 1800, 950, 2400, 3100, 1750, 4200, 2800, 1350, 3600])

total_sales = np.sum(sales)
average_sales = np.mean(sales)
median_sales = np.median(sales)
highest_sales = np.max(sales)
lowest_sales = np.min(sales)
std_sales = np.std(sales)
sales_above_average = sales[sales > average_sales]
p75_sales = np.percentile(sales, 75)

print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
print("Median Sales:", median_sales)
print("Highest Sale:", highest_sales)
print("Lowest Sale:", lowest_sales)
print("Standard Deviation:", std_sales)
print("Sales Above Average:", sales_above_average)
print("75th Percentile:", p75_sales)
```
