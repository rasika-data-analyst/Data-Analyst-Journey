## List, Set, Dict. Comprehension Assignment 

* Part 1: List Comprehensions


```python
#1 From a list of numbers [12, 45, 67, 89, 34, 22, 90, 77], extract all numbers greater than the average.

list = [12, 45, 67, 89, 34, 22, 90, 77]
avg = sum(list)/len(list)
results = [num for num in list if num > avg] 
print(results)
```

```python
#2 Create a list of tuples (num, square, cube) for numbers 1 to 15.
result2 = [(num, num**2, num**3) for num in range(1, 16)]
print(result2)
```

```python
#3 Given a string 'Python Comprehension Assignment', extract all words longer than 6 characters.
text = 'Python Comprehension Assignment'
results3 = [ word for word in text.split() if len(word) > 6]
print(results3)
```

```python
#4 Generate a list of numbers between 1–100 that are palindromes.
results4 = [num for num in range (1,101) if str(num)==str(num)[::-1]]
print(results4)
```

```python
#5 Convert a list ['apple', 'Banana', 'CHERRY', 'daTe'] into a new list where each word is
#titlecased and has length > 5. 

list  = ['apple', 'Banana', 'CHERRY', 'daTe']
new_list = [word for word in list if len(word) > 5]
print(new_list)
```

### Part 2: Dictionary Comprehensions


```python
#1 Create a dictionary mapping each number in range(1, 11) to 'Even' or 'Odd'.

result6 = {num: ('Even' if num % 2 == 0 else 'Odd') for num in range(1, 11)}
print(result6)
```

```python
#2 Given students = {'Vinit': 85, 'Anya': 92, 'Raj': 67, 'Sara': 78}, create a dictionary of students
#who scored above 80.

students = {'Vinit': 85, 'Anya': 92, 'Raj': 67, 'Sara': 78}
results7 = {name : marks for name, marks in students.items() if marks > 80}
print(results7)
```

```python
#3 Convert two lists keys = ['name', 'age', 'city', 'skills'] and values = ['Vinit', 26, 'Pune', ['Python',
#'SQL', 'Power BI']] into a dictionary using comprehension.

keys = ['name', 'age', 'city', 'skills']
values = ['Vinit', 26, 'Pune', ['Python','SQL', 'Power BI']]

results8 = {keys[i]: values[i] for i in range(len(keys))}
print(results8)
```

```python
#4 Create a dictionary from string 'intermediate' where key = letter, value = ASCII code.

text2 = 'intermediate'
results9 = {char: ord(char) for char in text2}
print(results9)
```

```python
#5 Reverse the key-value pair of {1:'a', 2:'b', 3:'c'} using dictionary comprehension. 

d = {1: 'a', 2: 'b', 3: 'c'}
results10 = {v: k for k, v in d.items()}
print(results10)
```

## Part 3: Set Comprehensions


```python
#1 From a list of words ['python', 'java', 'javascript', 'c++', 'python', 'java'], generate a set of unique words with length > 4.
list = ['python', 'java', 'javascript', 'c++', 'python', 'java']
set = {word for word in list if len(word)>4}
print(set)
```

```python
#2 From numbers 1–100, create a set of numbers divisible by 7 but not by 5.
set = {num for num in range (1, 101) if num%7 == 0 and num%5 != 0}
print(set)
```
