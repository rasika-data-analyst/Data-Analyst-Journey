```python
#43.	Question: Use a while loop to check if a list is sorted in ascending order. 

l1 = [20, 10, 40, 30, 50]

index = 0
is_sorted = True

while index < len(l1) - 1:
    if l1[index] > l1[index + 1]:
        is_sorted = False
        break
    index += 1

if is_sorted:
    print("List is sorted in ascending order")
else:
    print("List is NOT sorted")
```
