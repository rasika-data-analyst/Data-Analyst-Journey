```python
#1.	Write a program to take two numbers from the user and print the division result. Use exception handling to manage ZeroDivisionError.
try:
    a = int(input("Enter number 1: "))
    b = int(input("Enter number 2: "))
    print("Result:", a / b)
    
except ZeroDivisionError:
    print("Cannot divide by zero")
    
else: 
    print("Result: num is correct ", a / b)
    
```

```python
#2.	Take a value from the user and convert it into an integer. Handle ValueError if the input is not a number.

try:
    num = int(input("Enter a number: "))
    print("Number:", num)
except ValueError:
    print("Invalid input! Please enter a number")
```

```python
#3.	Create a program that tries to print the 5th index of a list. Handle IndexError.

try: 
    list = [10,20,30]
    print(list[5])
except IndexError:
    print("Index out of range")
        
```

```python
#4.	Write a program to open a file named data.txt in read mode. If the file is not found, handle FileNotFoundError.

try:
    f = open("data.txt", "r")
    print(f.read())
    f.close()
except FileNotFoundError:
    print("File not found")
```

```python
# 5.	Demonstrate use of finally block to print "Execution completed" whether exception occurred or not.

try:
    x = int(input("Enter number: "))
    print(10 / x)

except ZeroDivisionError:
    print("Cannot divide by zero")

except ValueError:
    print("Invalid input")

finally:
    print("Execution completed")
```

```python
#6.	Write a program with multiple exceptions: ZeroDivisionError, ValueError, IndexError.

try:
    a = int(input("Enter number: "))
    b = int(input("Enter number: "))
    lst = [1, 2, 3]
    
    print(a / b)
    print(lst[2])              # for indexerror put 5 above value

except ZeroDivisionError: 
    print("not divided by zero")
except ValueError:
    print("Enter the valid value")
except IndexError:
    print("Index is out of range")

finally:
    print("working fine")
    
```

```python
# 7.WAP to access the dictionary key. If the key is missing, handle KeyError.

try: 
    d = {"Name": "Ravi", "age" : 25}
    print(d["Salary"])
except KeyError:
    print("Key Not Found")
```

```python
# 8.Write a code that converts a string to integer inside try, if conversion succeeds print square of number using else.
try:
    a = int(input("Enter the num:   "))
except ValueError:
    print("Enter the valid num")
else :
    print("Square:" ,  a**2)
```

```python
#9.	Take input of two numbers and perform addition, subtraction, multiplication and division using one try block and appropriate except blocks.

try:
    a = int(input("Enter number1: "))
    b = int(input("Enter number2: "))
     
    print("Add", a+b)
    print("Sub", a-b)
    print("Multi", a*b)
    print("div", a/b)

except ZeroDivisionError:
    print("Cannot divide by zero")
except ValueError:
    print("Invalid input")
```

```python
#10.	Create a program for student result entry. If marks < 0 or > 100, raise InvalidMarksException.

class InvalidMarksException(Exception):
    pass

try:
    marks = int(input("Enter marks: "))
    
    if marks < 0 or marks > 100:
        raise InvalidMarksException("Marks should be between 0 and 100")
    
    print("Valid marks")

except InvalidMarksException as e:
    print(e)
    
```

```python
#11.	Write a file handling program:
#●	Ask user for a filename to open
#●	If file opens → display content
#●	If not → show custom message

filename = input("Enter filename: ")

try:
    f = open(filename, "r")
    print(f.read())
    f.close()
except FileNotFoundError:
    print("File does not exist. Please check the name.")
```

```python
# 12. Write a program to read data from a CSV file. Handle exceptions   like:
#●	File not found
#●	Permission denied
#●	Unicode decoding error

try:
    with open("data.csv", "r") as f:
        print(f.read())

except FileNotFoundError:
    print("File not found")
except PermissionError:
    print("Permission denied")
except UnicodeDecodeError:
    print("Encoding error while reading file")
```
