```python
import pandas as pd
import numpy as np

data = {
    'ID': [101, 102, 103, 104, 105, 106, 107, 108, 108, 109],
    'Name': ['Alice', 'Bob', 'Charlie', 'David', np.nan, 'Frank', 'Grace', 'Helen', 'Helen', 'Ian'],
    'Age': [25, 30, np.nan, 40, 45, -50, 55, 60, 60, 29],
    'Department': ['HR', 'IT', 'Finance', 'I.T.', 'Hr', 'Finance', 'IT', None, 'HR', 'IT'],
    'Salary': [50000, 60000, 75000, np.nan, 55000, 80000, 62000, 53000, 53000, 70000],
    'Joining Date': ['2020-05-10', '2018-08-15', '2016-09-25', '2017-11-30', '2019-07-01', np.nan, '2015-06-12', '2021-03-15', '2021-03-15', '2014-12-05']
}

df = pd.DataFrame(data)

print(df.head())
```

```python
#1.Fill missing values in categorical columns (like "Name" and "Department") with an appropriate value.

df["Name"].fillna("unknown", inplace = True)

df["Department"].fillna("IT", inplace = True)
```

```python
#2.Fill missing values in numerical columns (like "Age" and "Salary") using a suitable imputation technique.

df["Age"].fillna(df["Age"].mean(), inplace=True)
df['Salary'].fillna(df['Salary'].median(), inplace=True)
```

```python
#7. Standardize the department names (e.g., convert variations like "I.T.", "Hr" to a uniform format).

df["Department"]
```

```python
df['Department'] = df['Department'].str.upper().str.replace(".", "", regex=False)
```

```python
df["Department"]
```

```python
#Create a "Salary Category" column (e.g., Low, Medium, High based on salary ranges).

df["Salary"]
```

```python
def salary_category(salary):
    if pd.isna(salary):
        return "Unknown"
    elif salary < 60000:
        return "Low"
    elif salary < 75000:
        return "Medium"
    else:
        return "High"

```

```python
df['Salary_Category'] = df['Salary'].apply(salary_category)
```

```python
df['Salary_Category']
```

```python
df
```

```python
#Rename columns to have consistent naming conventions (e.g., "Joining Date" → "Joining_Date").
df.rename(columns={"Joining Date": "Joining_Date"}, inplace=True)

df.columns = df.columns.str.strip().str.replace(" ", "_")
```

```python
#13. Add a Bonus Column and the bonus will be calculated as 10% of the salary for all employees. If an employee has more than 5 years of experience, 
#they get an extra 5% bonus (total 15% of salary).

df['Experience'] = 2025 - df['Joining_Date'].dt.year
```

```python
df['Experience']
```

```python
def calculate_bonus(row):
    if row['Experience'] > 5:
        return row['Salary'] * 0.15   # 15%
    else:
        return row['Salary'] * 0.10   # 10%

df['Bonus'] = df.apply(calculate_bonus, axis=1)
```

```python
df['Bonus']
```

```python
df[['Salary', 'Experience', 'Bonus']]
```

```python
import numpy as np

df['Bonus'] = np.where(
    df['Experience'] > 5,
    df['Salary'] * 0.15,
    df['Salary'] * 0.10
)
```
